```python
pacientes = [
    {"idade": 68, "internacoes_anteriores": 2, "servico": "Cardiologia", "readmissao_30d": False},
    {"idade": 45, "internacoes_anteriores": 0, "servico": "Clínica Médica", "readmissao_30d": True},
    {"idade": 72, "internacoes_anteriores": 3, "servico": "Cardiologia", "readmissao_30d": True},
    {"idade": 51, "internacoes_anteriores": 1, "servico": "Neurologia", "readmissao_30d": False}
]
```

```python
for paciente in pacientes:
    print(paciente["idade"], paciente["servico"])
```

```python
for paciente in pacientes:
    if paciente["idade"] >= 65:
        print(f"Idade: {paciente['idade']} — 65 anos ou mais")
    else:
        print(f"Idade: {paciente['idade']} — menos de 65 anos")
```

```python
quantidade_readmissoes = 0

for paciente in pacientes:
    if paciente["readmissao_30d"]:
        quantidade_readmissoes = quantidade_readmissoes + 1

print(quantidade_readmissoes)
```

```python
soma_idades = 0

for paciente in pacientes:
    soma_idades = soma_idades + paciente["idade"]

print(soma_idades)
```

```python
media_idade = soma_idades / len(pacientes)
print(media_idade)
```

```python
pacientes = [
    {"id": "P01", "idade": 68, "internacoes_anteriores": 2, "servico": "Cardiologia", "readmissao_30d": False},
    {"id": "P02", "idade": 45, "internacoes_anteriores": 0, "servico": "Clínica Médica", "readmissao_30d": True},
    {"id": "P03", "idade": 72, "internacoes_anteriores": 3, "servico": "Cardiologia", "readmissao_30d": True},
    {"id": "P04", "idade": 51, "internacoes_anteriores": 1, "servico": "Neurologia", "readmissao_30d": False}
]

def atende_criterios_didaticos(paciente):
    return paciente["idade"] >= 65 and paciente["internacoes_anteriores"] >= 2

quantidade = 0

for paciente in pacientes:
    resultado = atende_criterios_didaticos(paciente)
    if resultado:
        quantidade += 1
    print(f"{paciente['id']}: {resultado}")

print(f"Total de registros que atendem aos critérios: {quantidade}")
```
