## Exercício 1 — Da lista para a tabela

Considere a seguinte lista de registros:

```python
pacientes = [
    {"id_paciente": 201, "idade": 29, "servico": "Ortopedia",   "dias_internacao": 3},
    {"id_paciente": 202, "idade": 71, "servico": "Cardiologia", "dias_internacao": 9},
    {"id_paciente": 203, "idade": 58, "servico": "Neurologia",  "dias_internacao": 5},
    {"id_paciente": 204, "idade": 83, "servico": "Cardiologia", "dias_internacao": 12},
    {"id_paciente": 205, "idade": 45, "servico": "Ortopedia",   "dias_internacao": 1}
]
```

Transforme essa lista em um DataFrame chamado `df`. Depois mostre o número de linhas e colunas, os nomes das colunas e o tipo de cada coluna.

### Resposta

````python
pacientes = [
    {"id_paciente": 201, "idade": 29, "servico": "Ortopedia",   "dias_internacao": 3},
    {"id_paciente": 202, "idade": 71, "servico": "Cardiologia", "dias_internacao": 9},
    {"id_paciente": 203, "idade": 58, "servico": "Neurologia",  "dias_internacao": 5},
    {"id_paciente": 204, "idade": 83, "servico": "Cardiologia", "dias_internacao": 12},
    {"id_paciente": 205, "idade": 45, "servico": "Ortopedia",   "dias_internacao": 1}
]

df = pd.DataFrame(pacientes)
df
````

````python
print(df.shape)     # (linhas, colunas)
print(df.columns)
print(df.dtypes)
````

Cada dicionário virou uma linha e cada chave virou uma coluna. `shape` retorna `(5, 4)`: 5 registros e 4 variáveis.
`id_paciente`, `idade` e `dias_internacao` são `int64`; `servico` é texto (`object`, ou `str` a partir do pandas 3.0).

## Exercício 2 — Selecionando partes da tabela

Usando o `df` do Exercício 1, escreva comandos para mostrar:

1. apenas as colunas `idade` e `servico`;
2. a primeira linha da tabela, usando `iloc`;
3. a idade do paciente no índice 3, usando `loc`.

### Resposta

````python
# 1. Duas colunas: colchetes duplos, porque passamos uma LISTA de nomes
df[["idade", "servico"]]
````

````python
# 2. Primeira linha, pela posição
df.iloc[0]
````

````python
# 3. Um valor, pelo rótulo da linha e nome da coluna
df.loc[3, "idade"]
````

- Várias colunas → o resultado é um **DataFrame**; uma coluna só (`df["idade"]`) → uma **Series**.
- `iloc` seleciona por **posição**; `loc` seleciona por **rótulo**. Como o índice é o automático (0, 1, 2…), aqui os dois coincidem.
- Resultado do item 3: `83`.

## Exercício 4 — Combinando condições e renomeando

Usando o `df`:

1. selecione os pacientes com **65 anos ou mais** e **mais de 10 dias** de internação;
2. crie um novo DataFrame `df_renomeado` em que a coluna `dias_internacao` passe a se chamar `tempo_internacao`, e mostre as colunas dele.

### Resposta

````python
# Cada condição entre parênteses, unidas por & ("e")
df[(df["idade"] >= 65) & (df["dias_internacao"] > 10)]
````

````python
df_renomeado = df.rename(columns={"dias_internacao": "tempo_internacao"})
df_renomeado.columns
````

````python
# O df original continua igual: a renomeação foi guardada em um novo objeto
df.columns
````

- Só o paciente **204** atende às duas condições (83 anos, 12 dias).
- Esquecer os parênteses em cada condição é um dos erros mais comuns.
- `rename()` muda só o nome da coluna, não os valores.

## Exercício 5 (opcional) — Diagnóstico

Considere o seguinte DataFrame:

```python
df_teste = pd.DataFrame({
    "idade":   [68, 45, "setenta e dois", 51],
    "servico": ["Cardiologia", "cardiologia", "Neurologia", "Cardiologia "]
})
```

Use `dtypes` e `unique()` para encontrar **dois problemas** nessa base. Seria seguro calcular a média de idade diretamente?

### Resposta

````python
df_teste = pd.DataFrame({
    "idade":   [68, 45, "setenta e dois", 51],
    "servico": ["Cardiologia", "cardiologia", "Neurologia", "Cardiologia "]
})

df_teste.dtypes
````

````python
df_teste["servico"].unique()
````

**Problema 1 — tipo:** `idade` aparece como `object` porque um valor está escrito por extenso ("setenta e dois"). Não é seguro calcular a média: a coluna precisa ser tratada antes.

**Problema 2 — categorias:** `unique()` mostra três grafias para o mesmo serviço: `"Cardiologia"`, `"cardiologia"` e `"Cardiologia "` (com espaço no final). Para o pandas, são três categorias diferentes.

O problema não está no código, e sim na forma como os dados foram registrados. **É exatamente isso que vamos resolver hoje, na Aula 5: limpeza e padronização.**
