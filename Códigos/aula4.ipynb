## De registros para tabelas

```python
pacientes = [
    {"id_paciente": 101, "idade": 68, "internacoes_anteriores": 2,
     "servico": "Cardiologia", "readmissao_30d": False},
    {"id_paciente": 102, "idade": 45, "internacoes_anteriores": 0,
     "servico": "Clínica Médica", "readmissao_30d": True},
    {"id_paciente": 103, "idade": 72, "internacoes_anteriores": 3,
     "servico": "Cardiologia", "readmissao_30d": True},
    {"id_paciente": 104, "idade": 51, "internacoes_anteriores": 1,
     "servico": "Neurologia", "readmissao_30d": False}
]
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(pacientes)
df
```

```python
print(type(df))
print(type(df["idade"][0]))
```

## Escrita e Leitura de arquivos


Escrevendo um csv

```python
df.to_csv("pacientes_aula4.csv", index=False)
```

Escrevendo um excel

```python
df.to_excel("pacientes_aula4.xlsx", index=False)
```

Lendo um arquivo csv

```python
df_csv = pd.read_csv("pacientes_aula4.csv")
df_csv
```

Lendo um arquivo excel

```python
df_excel = pd.read_excel("pacientes_aula4.xlsx")
df_excel
```
